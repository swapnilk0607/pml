```python
import cv2
import numpy as np

# Load the image
image_path = "img.png"   # change this path
img = cv2.imread(image_path)

# Get image dimensions
h, w, _ = img.shape

# Define number of grid rows and columns
rows, cols = 8, 8
tile_h, tile_w = h // rows, w // cols

tiles = []

# Divide the image into 8x8 grid
for i in range(rows):
    for j in range(cols):
        y1 = i * tile_h
        y2 = (i + 1) * tile_h
        x1 = j * tile_w
        x2 = (j + 1) * tile_w

        tile = img[y1:y2, x1:x2]
        tiles.append(tile)

        # optional: save each tile
        success =cv2.imwrite(f"tile_{i}_{j}.png", tile)
        print(f"Succeeded: {success}")

print(f"Total tiles created: {len(tiles)}")

# Show an example tile
# cv2.imshow("Example Tile", tiles[20])
# cv2.waitKey(0)
# cv2.destroyAllWindows()

```
