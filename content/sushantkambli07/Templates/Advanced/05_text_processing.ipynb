```python
# Dense embeddings (e.g., averaged Word2Vec vectors) -> numpy
import numpy as np
from sklearn.metrics.pairwise import cosine_distances

# example dense vectors (replace with actual doc vectors)
v1 = np.array([0.1, 0.3, -0.2])
v2 = np.array([0.0, 0.25, -0.1])

# manual cosine similarity
cos_sim = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
cos_dist = 1.0 - cos_sim

# or via sklearn
cos_dist_sklearn = cosine_distances(v1.reshape(1,-1), v2.reshape(1,-1))[0,0]

print("cosine similarity:", cos_sim)
print("cosine distance:", cos_dist, "(sklearn:", cos_dist_sklearn, ")")
```
